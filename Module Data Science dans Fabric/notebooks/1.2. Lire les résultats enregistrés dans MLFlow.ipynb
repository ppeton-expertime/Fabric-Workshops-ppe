# 1.2. Lire les résultats enregistrés dans MLflow
Ce notebook permet de consulter les runs MLflow, leurs métriques, leurs paramètres et les artefacts associés au modèle de prévision vélo.

## Paramétrage commun
Ce notebook lit par défaut les runs de la V1 baseline. Modifier la variable d'expérience si vous souhaitez inspecter la V2 FLAML ou la V3 SynapseML.

```python
import mlflow
import pandas as pd
```

```python
registered_model_name = "prediction_velos_horaires"
selected_experiment_name = "ds_prediction_velos_v1"
print({"registered_model_name": registered_model_name, "selected_experiment_name": selected_experiment_name})
```

```python
experiment_name = selected_experiment_name
experiment = mlflow.get_experiment_by_name(experiment_name)
if experiment is None:
    raise ValueError(f"Experience introuvable: {experiment_name}")
runs = mlflow.search_runs(experiment_ids=[experiment.experiment_id])
display(runs)
```

```python
metrics_cols = [col for col in runs.columns if col.startswith("metrics.")]
params_cols = [col for col in runs.columns if col.startswith("params.")]
display(runs[["run_id", "start_time"] + metrics_cols + params_cols])
```

## Comparer les runs
Cette étape affiche uniquement les colonnes utiles pour comparer les métriques et les paramètres entre plusieurs runs.

```python
best_run_id = runs.sort_values(by="metrics.rmse", ascending=True).iloc[0]["run_id"]
best_run = mlflow.get_run(best_run_id)
print("Run retenu :", best_run_id)
print("Params :", best_run.data.params)
print("Metrics :", best_run.data.metrics)
```

```python
artifact_root = best_run.info.artifact_uri
print("Artifact URI du run retenu :", artifact_root)
print("Utiliser le volet MLflow de Fabric pour naviguer visuellement dans les artefacts si necessaire.")
```

## À retenir
MLflow permet de comparer rapidement les runs d'une même expérience, de retrouver les paramètres, les métriques et les artefacts du modèle retenu.
