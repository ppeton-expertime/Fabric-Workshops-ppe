# 0.1. Télécharger le fichier CSV et le déposer dans un Lakehouse Fabric
Ce notebook sert de support opératoire pour récupérer le fichier CSV source et le déposer dans le Lakehouse du workspace [FORMATION].

## Étapes manuelles
1. Télécharger le fichier depuis https://www.data.gouv.fr/api/1/datasets/r/237382af-0e7a-4ef8-9508-b3e9e78adcfd.
2. Ouvrir Microsoft Fabric et le workspace [FORMATION].
3. Ouvrir le Lakehouse cible.
4. Aller dans Files puis Upload > Files.
5. Déposer le fichier CSV téléchargé dans le Lakehouse.

```python
source_file_name = "comptage-velo-donnees-compteurs.csv"
lakehouse_target_path = f"Files/{source_file_name}"
print("Fichier attendu dans le Lakehouse :", lakehouse_target_path)
```

```python
try:
    df = spark.read.option("header", True).option("sep", ";").csv(lakehouse_target_path)
    print("Lecture du fichier reussie")
    display(df.limit(10))
except Exception as exc:
    print("Le fichier n'est pas encore present dans le Lakehouse ou le nom doit etre adapte.")
    print(exc)
```

## À retenir
Le fichier source doit être disponible dans la zone Files du Lakehouse avant de pouvoir être exploité dans les notebooks suivants.
