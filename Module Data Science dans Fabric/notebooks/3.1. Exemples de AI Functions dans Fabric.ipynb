# 3.1. Exemples de AI Functions dans Fabric
Ce notebook regroupe plusieurs démonstrations d'AI Functions en PySpark et en SQL, adaptées du contenu Microsoft Learn.

## Référentiel de l'atelier
Les exemples de ce notebook reprennent les grands cas d'usage documentés dans Microsoft Learn : sentiment, classification, extraction, résumé, SQL et multimodal.

```python
import synapse.ml.spark.aifunc as aifunc
```

## Exemple 1 : analyser le sentiment
On commence par un dataframe simple de commentaires libres, puis on ajoute une colonne de sentiment.

```python
reviews_df = spark.createDataFrame([
    ("Le service etait excellent et rapide.",),
    ("La piste cyclable etait mal entretenue et dangereuse.",),
    ("Le trajet etait correct, sans plus.",)
], ["avis"])
sentiment_df = reviews_df.ai.analyze_sentiment(input_col="avis", output_col="sentiment")
display(sentiment_df)
```

## Exemple 2 : classifier et extraire
Les cellules suivantes montrent comment catégoriser un texte puis extraire des champs structurés.

```python
classification_df = spark.createDataFrame([
    ("Retard important sur la ligne de bus ce matin.",),
    ("Station velo propre et bien positionnee.",),
    ("Application mobile difficile a utiliser.",)
], ["commentaire"])
classified = classification_df.ai.classify(labels=["transport", "station", "application", "autre"], input_col="commentaire", output_col="categorie")
display(classified)
```

```python
extract_df = spark.createDataFrame([
    ("Marie Dupont habite a Lyon et travaille comme data analyst.",),
    ("Karim Benali vit a Lille et exerce comme chef de projet.",)
], ["description"])
entities = extract_df.ai.extract(labels=["nom", "ville", "profession"], input_col="description")
display(entities)
```

```python
summary_df = spark.createDataFrame([
    ("Le service de location de velos a connu une forte hausse d'utilisation ce week-end autour des zones touristiques.",)
], ["texte"])
summaries = summary_df.ai.summarize(input_col="texte", output_col="resume")
display(summaries)
```

## Exemple 3 : résumer, puis passer en SQL et en multimodal
La fin du notebook enchaîne un résumé PySpark, une requête SQL Fabric et un exemple de fichier traité comme entrée.

```python
WITH df AS (
    SELECT *
    FROM ( VALUES
        ('La borne est hors service depuis hier.'),
        ('Le parcours est agreable et fluide.'),
        ('Le plan des stations est peu lisible.')
    ) AS df(commentaire)
)
SELECT
    commentaire,
    AI_CLASSIFY(commentaire, 'incident', 'experience', 'signaletique', 'autre') AS categorie
FROM df;
```

```python
file_df = spark.createDataFrame([
    ("Files/docs/rapport-stations.pdf",)
], ["file_path"])
multimodal_summary = file_df.ai.summarize(input_col="file_path", input_col_type="path", output_col="summary")
display(multimodal_summary)
```
